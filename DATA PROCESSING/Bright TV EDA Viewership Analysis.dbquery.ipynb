```sql

----------------------------------------------------------------------
                  --BRIGHT TV USER VIEWERSHIP ANALYSIS--
----------------------------------------------------------------------
--USER PROFILES--

-----------------------TABLE OVERVIEW---------------------------------
--The following SYNTAXs will shows the entire structure of the table before any analysis can be done.
--Shows only Total Number of COLUMNS
DESCRIBE bright_tv.default.bright_tv_user_profiles;

--Shows the ENTIRE TABLE inclueding COLUMNS and ROWS
SELECT *
FROM bright_tv.default.bright_tv_user_profiles;

--The following SYNTAX works the same as the above one. However, we have LIMITED rows to 100.
SELECT *
FROM bright_tv.default.bright_tv_user_profiles
LIMIT 100;

--Checking how big the data is. (TOTAL RECORDS)
SELECT COUNT(*) AS Total_Records
FROM bright_tv.default.bright_tv_user_profiles;

--Checking for DUPLICATES by USERID
SELECT
    COUNT(*) AS Total_Records,
    COUNT(DISTINCT UserID) AS Number_of_Subscribers,
    COUNT(*) - COUNT(DISTINCT UserID) AS Duplicate_Records
FROM bright_tv.default.bright_tv_user_profiles;

--NOTE:The User Profiles table contains 5,375 subscriber records and 5,375 unique subscribers. Since the number of records equals the number of unique UserIDs, there are no duplicate subscriber records in the dataset. This confirms that the User Profiles table is clean and each subscriber is represented only once.

                      -------------------CHECKING FOR NULL VALUES---------------------
--We are now checking if we have NULL values within each COLUMNS. Basically the SYNTAX checks each row for NULL values.
--NULL values are placeholders that represent that information is missing and may be added at a later stage and not that it means a ZERO. They do not represent a zero.

SELECT
    COUNT(CASE WHEN UserID IS NULL THEN 1 END) AS UserID_Nulls,
    COUNT(CASE WHEN Name IS NULL THEN 1 END) AS Name_Nulls,
    COUNT(CASE WHEN Surname IS NULL THEN 1 END) AS Surname_Nulls,
    COUNT(CASE WHEN Email IS NULL THEN 1 END) AS Email_Nulls,
    COUNT(CASE WHEN Gender IS NULL THEN 1 END) AS Gender_Nulls,
    COUNT(CASE WHEN Race IS NULL THEN 1 END) AS Race_Nulls,
    COUNT(CASE WHEN Age IS NULL THEN 1 END) AS Age_Nulls,
    COUNT(CASE WHEN Province IS NULL THEN 1 END) AS Province_Nulls,
    COUNT(CASE WHEN `Social Media Handle` IS NULL THEN 1 END) AS SocialMedia_Nulls
FROM bright_tv.default.bright_tv_user_profiles;

--Outcome: There are no NULL values within each columns.

                   -------------------CHECKING FOR ROWS THAT DO NOT HAVE VALUES---------------------
SELECT
    SUM(CASE WHEN Name = ' ' THEN 1 ELSE 0 END) AS Name_Empty_Count,
    SUM(CASE WHEN Surname = ' ' THEN 1 ELSE 0 END) AS Surname_Empty_Count,
    SUM(CASE WHEN Email = ' ' THEN 1 ELSE 0 END) AS Email_Empty_Count,
    SUM(CASE WHEN Gender = ' ' THEN 1 ELSE 0 END) AS Gender_Empty_Count,
    SUM(CASE WHEN Race = ' ' THEN 1 ELSE 0 END) AS Race_Empty_Count,
    SUM(CASE WHEN Province = ' ' THEN 1 ELSE 0 END) AS Province_Empty_Count,
    SUM(CASE WHEN `Social Media Handle` = ' ' THEN 1 ELSE 0 END) AS SocialMediaHandle_Empty_Count
FROM bright_tv.default.bright_tv_user_profiles;
--The Syntax checks for empty spaces in selected columns. We then use THE CASE Statement to identify which records are empty and assigning a value of 1 for empty fields and 0 for fields that contain data. The SUM() function then adds these values together to calculate the total number of empty records for each column, while the AS statement gives each result a clear name. In simple terms, it creates a data quality report showing how many missing/blank values exist in each column, which helps a Data Analyst identify cleaning issues before analysing the data.

        ------------------------CHECKING FOR ROWS WITH 'NONE' VALUES-----------------------------
SELECT
    SUM(CASE WHEN UserID ILIKE 'None' THEN 1 ELSE 0 END) AS UserID_None,
    SUM(CASE WHEN Name ILIKE 'None' THEN 1 ELSE 0 END) AS Name_None,
    SUM(CASE WHEN Surname ILIKE 'None' THEN 1 ELSE 0 END) AS Surname_None,
    SUM(CASE WHEN Gender ILIKE 'None' THEN 1 ELSE 0 END) AS Gender_None,
    SUM(CASE WHEN Age ILIKE 'None' THEN 1 ELSE 0 END) AS Age_None,
    SUM(CASE WHEN Email ILIKE 'None' THEN 1 ELSE 0 END) AS Email_None,
    SUM(CASE WHEN Province ILIKE 'None' THEN 1 ELSE 0 END) AS Province_None,
    SUM(CASE WHEN Race ILIKE 'None' THEN 1 ELSE 0 END) AS Race_None,
    SUM(CASE WHEN `Social Media Handle` ILIKE 'None' THEN 1 ELSE 0 END) AS SocialMedialHandle_None
FROM bright_tv.default.bright_tv_user_profiles;
--The ILIKE OPERATOR goes to each row and returns each column with the total amount of values containing a 'NONE' value.

----------------USERID--------------
SELECT
    COUNT(UserID) AS Total_Valid_Users
FROM bright_tv.default.bright_tv_user_profiles
WHERE UserID NOT LIKE 'None';
--It removes all 'NONE' values within the USERID column and returns the Total Number Of Suscribers which is  5375 
--Meaning there are no 'NONE'.

----------------NAME--------------
SELECT Name
FROM bright_tv.default.bright_tv_user_profiles;

SELECT
    COUNT(Name) AS Total_None_Names
FROM bright_tv.default.bright_tv_user_profiles
WHERE Name LIKE 'None';
--Returns 920 INACTIVE USERS
SELECT
    COUNT(Name) AS Total_Valid_Names
FROM bright_tv.default.bright_tv_user_profiles
WHERE Name NOT LIKE 'None';
--ReturnS 4455 ACTIVE USERS
----------------SURNAME--------------
SELECT Surname
FROM bright_tv.default.bright_tv_user_profiles;

SELECT
    COUNT(Surname) AS Total_None_Surnames
FROM bright_tv.default.bright_tv_user_profiles
WHERE Surname LIKE 'None';
--Returned 920
SELECT
    COUNT(Surname) AS Total_Valid_Surnames
FROM bright_tv.default.bright_tv_user_profiles
WHERE Surname NOT LIKE 'None';
--Returned 4455

----------------GENDER--------------
--Checking for number of gender
--Types of Genders
SELECT DISTINCT Gender
FROM bright_tv.default.bright_tv_user_profiles;
-- Returns the Types of Genders we have in the Gender Column
SELECT
    Gender,
    COUNT(*) AS Total_Users_Per_Gender
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY Gender
ORDER BY Total_Users_Per_Gender DESC;
--Shows the total number of Genders withing the Gender Column 
--Males - 3918
--Females - 537
--None - 702
--Empty - 218
--NOTE: Replace 'None' & ' ' with suitable values

----------------PROVINCE--------------
--Checking how many provinces we have 
SELECT DISTINCT Province
FROM bright_tv.default.bright_tv_user_profiles;

--Provinces and the Total_Users_Per_Province
SELECT
    Province,
    COUNT(*) AS Total_Users_Per_Province
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY Province
ORDER BY Total_Users_Per_Province DESC;
--Returns each Province and the Total Number of each Subscriber per Provinve
Gauteng	- 1704
Western Cape - 791
None - 702
Kwazulu Natal -	481
Mpumalanga - 421
Limpopo	- 368
Eastern Cape - 287
' '  - 218
North West - 161
Free State - 156
Northern Cape - 86
--NOTE:Clean and replace 'NONE' and 'EMPTY' values with 'UNCATEGORISED'
----------------RACE--------------
--Checking the Types and number of RACE
--Types of Race
SELECT DISTINCT Race
FROM bright_tv.default.bright_tv_user_profiles;

--Returns Total Number of Subscribers Per Race
SELECT 
        COUNT(DISTINCT Race) AS Number_of_Subs_Per_Races
FROM bright_tv.default.bright_tv_user_profiles;

--Types of Race and their total nmuber of users per race
SELECT
    Race,
    COUNT(*) AS Total_Users_Per_Race
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY Race
ORDER BY total_users_per_race DESC;

--Black - 1811
--'None' - 1078
--Indian_asian - 768
--White - 760
--Coloured - 679
--' '  - 231
--Other - 48
--NOTE: Clean and replace 'None' and ' ' values with 'Unknown'
----------------AGE--------------
--NOTE: The AGE column does not contain NULLs,None nor Empty values
--Therefore we will only attend to it when we start cleaning the data
SELECT Age 
FROM bright_tv.default.bright_tv_user_profiles
LIMIT 100;

SELECT MAX(Age) AS max_age,
       MIN(Age) AS min_age
FROM bright_tv.default.bright_tv_user_profiles;
--Returns Max and Min age
--MIN - 0
--MAX - 114
SELECT
    CASE
        WHEN Age = 0 THEN 'Infant'
        ELSE Gender
    END AS Gender,Age
FROM bright_tv.default.bright_tv_user_profiles;
--Putting INFANT in each row that contains 0 in the Age column.
--NOTE USE THE ABOVE WHEN 'JOINING' the table

SELECT
        COUNT(DISTINCT UserID) AS subs,
    CASE
        WHEN Age BETWEEN 0 AND 2 THEN 'Infants'
        WHEN Age BETWEEN 3 AND 12 THEN 'Kids'
        WHEN Age BETWEEN 13 AND 19 THEN 'Teens'
        WHEN Age BETWEEN 20 AND 35 THEN 'Young Adults'
        WHEN Age BETWEEN 36 AND 50 THEN 'Adults'
        WHEN Age BETWEEN 51 AND 64 THEN 'Mature Adults'
        WHEN Age >= 65 THEN 'Seniors'       
    END AS Age_Bucket
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY Age_Bucket;
--NOTE: Age Buckets will be used when JOINING the two tables and in visualizing.
----------------SOCIAL MEDIA HANDLE DATA SCREENING--------------
SELECT `Social Media Handle`
FROM bright_tv.default.bright_tv_user_profiles;

SELECT
    COUNT(`Social Media Handle`) AS Total_Valid_Handles
FROM bright_tv.default.bright_tv_user_profiles
WHERE `Social Media Handle` LIKE 'None';
--Returned 920

SELECT
    COUNT(`Social Media Handle`) AS Total_Valid_Handles
FROM bright_tv.default.bright_tv_user_profiles
WHERE `Social Media Handle` NOT LIKE 'None';
--Returned 4455

SELECT
    COUNT(Email) AS Total_Valid_Emails
FROM bright_tv.default.bright_tv_user_profiles
WHERE Email NOT LIKE 'None';

----------------EMAIL--------------
SELECT Email
FROM bright_tv.default.bright_tv_user_profiles;

SELECT
    COUNT(Email) AS Total_None_Email
FROM bright_tv.default.bright_tv_user_profiles
WHERE Email LIKE 'None';
--Returned 0
SELECT
    COUNT(Email) AS Total_Valid_Emails
FROM bright_tv.default.bright_tv_user_profiles
WHERE Email NOT LIKE 'None';
--Returned 5375
--NOTE: EMAILS and Social Media Handles will be used in the actual presentation, for suggestions.

----------------------------------------------------------------------
                              --DATA CLEANING--
----------------------------------------------------------------------
--Above we were exploring the data and now we will be commencing to cleaning the data as we now know what we are dealing with.
--REPLACING NONE VALUES IN THE NAME & SURNAME COLUMNS WITH WORD 'UNKNOWN'
SELECT
    CASE
        WHEN Name ILIKE 'None' THEN 'Unknown'
        ELSE Name
    END AS Name,

    CASE
        WHEN Surname ILIKE 'None' THEN 'Unknown'
        ELSE Surname
    END AS Surname
FROM bright_tv.default.bright_tv_user_profiles;

--REPLACING 'NONE' & ' ' VALUES IN THE GENDER COLUMN WITH WORD 'UNKNOWN'
SELECT COUNT(*) AS Count,
       COUNT(DISTINCT UserID) AS Subscribers,
    CASE 
        WHEN Gender = ' ' THEN 'Unknown'
        WHEN Gender = 'None' THEN 'Unknown'
        WHEN Gender iLIKE 'Unknown' THEN 'Unknown' ELSE Gender END AS Gender
FROM bright_tv.default.bright_tv_user_profiles

GROUP BY 
    CASE 
        WHEN Gender = ' ' THEN 'Unknown'
        WHEN Gender = 'None' THEN 'Unknown'
        WHEN Gender iLIKE 'Unknown' THEN 'Unknown' ELSE Gender END;

--NOTES FOR THE ABOVE SYNTAX--
--STEP 1: Read the Gender column
--The query starts by looking at every record in the Gender column of the bright_tv_user_profiles table. Its goal is to understand what gender values exist and how many users belong to each gender category.

--STEP 2: Clean the Gender values using the CASE statement
--The CASE statement checks each Gender value and looks for records that contain an empty space (' '), None, or Unknown. Instead of treating these as separate categories, it changes all of them to a single value called "Unknown". This is done to improve data quality and ensure that missing or unclear gender information is grouped together.

--STEP 3: Keep valid Gender values unchanged
--If a Gender value is not an empty space, None, or Unknown, the query leaves it exactly as it is. For example, values such as Male and Female remain unchanged because they are valid gender entries.

--STEP 4: Count the total number of records
--The statement: COUNT(*)
--Counts all records within each gender group. This tells us how many rows belong to each category after the Gender values have been cleaned.

--STEP 5: Count the number of unique subscribers
--The statement: COUNT(DISTINCT UserID)
--Counts the number of unique users in each gender group. This prevents the same user from being counted multiple times if they appear more than once in the dataset.

--STEP 6: Group similar Gender values together
--The GROUP BY statement uses the same CASE logic that was used to clean the Gender column. --This ensures that all records labelled as empty spaces, None, or Unknown are placed into a single group called Unknown instead of being counted separately.

--Rofhiwa's Syntax
--When i apply Rofhiwa's syntax it still returns a NONE and not UNKNOWN
SELECT COUNT(*) AS Count,
       COUNT(DISTINCT UserID) AS Subscribers,
    CASE 
        WHEN gender = ' ' THEN 'Unknown'
        WHEN gender ILIKE 'Unknown' THEN 'Unknown' 
        ELSE gender 
        END AS Gender
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY Gender;

--NOTE: LIKE is case sensitive and ILIKE is not case sensitive
--LIKE will return values according to a certain pattern. For example if you are looking WHERE Gender LIKE 'male' in small letters, it will return only male values with small letters and leave those with capital letters and others such as 'MALE' and 'Male'
--ILIKE on the other hand is not case sensitive, if you use WHERE Gender ILIKE 'male', it will return all gender values as they are male,Male,MALE
--Changinging values under the Gender column to start with a Capital letter using ILIKE Function
SELECT
    CASE
        WHEN Gender ILIKE 'female' THEN 'Female'
        WHEN Gender ILIKE 'male' THEN 'Male'
        ELSE Gender
    END AS Gender
FROM bright_tv.default.bright_tv_user_profiles;
--NOTE: Use the above when combining tables

SELECT Gender
FROM bright_tv.default.bright_tv_user_profiles
WHERE Gender ILIKE 'male';
--Checkicng the structure of the values.
--Note: All MALES and FEMALES start with a lower cases

---------------PROVINCE-------------------
--Removing NONE and Empty values under PROVINCE

SELECT DISTINCT
    CASE 
        WHEN Province = ' ' THEN 'Uncategorised'
        WHEN Province = 'None' THEN 'Uncategorised'
        WHEN Province iLIKE 'Uncategorised' THEN 'Uncategorised' ELSE Province END AS Region
FROM bright_tv.default.bright_tv_user_profiles;
--ALTERNATIVE SYNTAX--
SELECT 
        COUNT(*) AS Count,
        COUNT(DISTINCT Province) AS Total_Number_Per_Province,
    CASE 
        WHEN Province = ' ' THEN 'Uncategorised'
        WHEN Province = 'None' THEN 'Uncategorised'
        WHEN Province ILIKE 'Uncategorised' THEN 'Uncategorised' 
        ELSE Province 
        END AS Region
FROM bright_tv.default.bright_tv_user_profiles
GROUP BY
 CASE 
        WHEN Province = ' ' THEN 'Uncategorised'
        WHEN Province = 'None' THEN 'Uncategorised'
        WHEN Province ILIKE 'Uncategorised' THEN 'Uncategorised' 
        ELSE Province 
        END;
--Returns Preovice with their active subscribers pe province inclueding Uncategorised ones.
--NOTE: Use the above SYNTAX when combining tables
--------------------ALTERNATIVE SYNTAX-------------------------

SELECT DISTINCT
    CASE 
        WHEN Province = ' ' OR Province = 'None' THEN 'Uncategorised'
        WHEN Province ILIKE 'Uncategorised' THEN 'Uncategorised' 
        ELSE Province 
        END AS Region
FROM bright_tv.default.bright_tv_user_profiles;

----------------RACE-------------------------
--Repalcing 'NONE' and 'EMPTY' values 

SELECT DISTINCT
    CASE 
        WHEN Race = ' ' THEN 'Unknown'
        WHEN Race ILIKE 'None' THEN 'Unknown'
        WHEN Race ILIKE 'Other' THEN 'Unknown'
        WHEN RAce ILIKE 'Uunknown' THEN 'Unknown' 
        ELSE Race 
        END AS Race
FROM bright_tv.default.bright_tv_user_profiles;
---------------TOTAL NUMBER OF ACTIVE SUBSCRIBERS PER RACE---------------------
SELECT 
    COUNT(*) AS Count,
    COUNT(DISTINCT UserID) AS Total_Subs_Per_Race,

    CASE
        WHEN Race = ' ' THEN 'Unknown'
        WHEN Race ILIKE 'None' THEN 'Unknown'
        WHEN RACE ILIKE 'Other' THEN 'Unknown'
        WHEN Race ILIKE 'Unknown' THEN 'Unknown'
        ELSE Race
    END AS Race

FROM bright_tv.default.bright_tv_user_profiles

GROUP BY
    CASE
        WHEN Race = ' ' THEN 'Unknown'
        WHEN Race ILIKE 'None' THEN 'Unknown'
        WHEN Race ILIKE 'Other' THEN 'Unknown'
        WHEN Race ILIKE 'Unknown' THEN 'Unknown'
        ELSE Race
    END
ORDER BY Total_Subs_Per_Race DESC;
--Note: Use it when combining tables

-------------------SOCIAL MEDIAL HANDLE------------------------
SELECT
    CASE
        WHEN `Social Media Handle` ILIKE 'None' THEN 'No Social Media Handle'
        ELSE `Social Media Handle`
    END AS Social_Media_Handle
FROM bright_tv.default.bright_tv_user_profiles;
--NOTE: USE THE ABOVE SYNTAX WHEN JOINING TABLES


WITH cte1 AS (
SELECT UserID,
    CASE
        WHEN Province = ' ' THEN 'Uncategorised'
        WHEN Province ILIKE 'None' THEN 'Uncategorised'
        WHEN Province ILIKE 'Uncategorised' THEN 'Uncategorised' 
        ELSE Province 
    END AS Region,

        Age,
    CASE
        WHEN Age BETWEEN 0 AND 2 THEN 'Infants'
        WHEN Age BETWEEN 3 AND 12 THEN 'Kids'
        WHEN Age BETWEEN 13 AND 19 THEN 'Teens'
        WHEN Age BETWEEN 20 AND 35 THEN 'Young Adults'
        WHEN Age BETWEEN 36 AND 50 THEN 'Adults'
        WHEN Age BETWEEN 51 AND 64 THEN 'Mature Adults'
        WHEN Age >= 65 THEN 'Seniors'       
    END AS Age_Bucket,

       CASE
        WHEN Race = ' ' THEN 'Unknown'
        WHEN Race = 'black' THEN 'Black'
        WHEN Race = 'white' THEN 'White'
        WHEN Race = 'indian_asian' THEN 'Indian_Asian'
        WHEN Race = 'coloured' THEN 'Coloured'
        WHEN Race ILIKE 'None' THEN 'Unknown'
        WHEN RACE ILIKE 'Other' THEN 'Unknown'
        WHEN Race ILIKE 'Unknown' THEN 'Unknown'
        ELSE Race
    END AS Race,

    CASE 
        WHEN gender = ' ' THEN 'Unknown'
        WHEN gender = 'female' THEN 'Female'
        WHEN gender = 'male' THEN 'Male'
        WHEN gender ILIKE 'None' THEN 'Unknown'
        WHEN gender ILIKE 'Unknown' THEN 'Unknown' 
        ELSE gender 
    END AS Gender,
       
    CASE
         WHEN Email IS NOT NULL OR Email = ' ' OR Email NOT IN ('None') THEN 1 
         ELSE 0
        END AS email_flag,
       
    CASE
         WHEN `Social Media Handle` IS NOT NULL OR `Social Media Handle` = ' ' OR `Social Media Handle` NOT IN ('None') THEN 1 
         ELSE 0
        END AS social_media_flag
FROM bright_tv.default.bright_tv_user_profiles
)
SELECT*
FROM cte1;

---------------------------------------------------------------------
                --VIEWERSHIP DATA ANALYSIS--
---------------------------------------------------------------------
-----------------------TABLE OVERVIEW--------------------------------
SELECT *
FROM bright_tv.default.bright_tv_viewership
LIMIT 10;

--TOTAL NUMBER OF SUBSCRIBERS
SELECT COUNT(*) AS number_of_rows,
       COUNT(COALESCE(UserID0,userid4)) AS Active_Subscribers
FROM bright_tv.default.bright_tv_viewership;
--NOTE: There are 10000 subscribers

--ACTIVE SUBSCRIBERS
SELECT COUNT(*) AS number_of_rows,
       COUNT(COALESCE(UserID0,userid4)) AS Active_Subscribers,
       COUNT(DISTINCT COALESCE(UserID0,userid4)) AS Active_Users
FROM bright_tv.default.bright_tv_viewership;
--NOTE: From 10000 subscribers, only 4386 are active

--CHECKING FOR ROWS WITH 'NO NULL' VALUES 
SELECT COUNT(*) AS number_of_rows,
       COUNT(COALESCE(UserID0,userid4)) AS Active_Subscribers,
       COUNT(DISTINCT COALESCE(UserID0,userid4)) AS Active_Users
FROM bright_tv.default.bright_tv_viewership
WHERE UserID0 IS NOT NULL OR userid4 IS NOT NULL;
--NOTE: There are no NULL values

--CHECKING FOR ROWS WITH 'NULL' VALUES
SELECT COUNT(*)
FROM bright_tv.default.bright_tv_viewership
WHERE UserID0 IS NULL
OR Channel2 IS NULL
OR RecordDate2 IS NULL
OR `Duration 2` IS NULL
OR userid4 IS NULL;
--NOTE: There are no NULL, NONE nor '' values in each column.

----------------------CHANNELS---------------------------
SELECT DISTINCT Channel2 
FROM bright_tv.default.bright_tv_viewership;
--Note: We have 21 different channels

--CHECKING FOR '', 'None'& 'Other'
SELECT 
    SUM(CASE WHEN Channel2 = ' ' THEN 1 ELSE 0 END) AS Channel_Empty_Count,
    SUM(CASE WHEN Channel2 = 'None' THEN 1 ELSE 0 END) AS None_Count,
    SUM(CASE WHEN Channel2 = 'Other' THEN 1 ELSE 0 END) AS Other_Count
FROM bright_tv.default.bright_tv_viewership;
--Note: No empty, none nor other values werelocated

--------------------RECORD DATE------------------
SELECT RecordDate2
FROM bright_tv.default.bright_tv_viewership;

--EXTRACTING YEAR, MONTH & DAY FROM RECORDs COLUMN
SELECT 
    RecordDate2,
    EXTRACT(YEAR FROM CAST(RecordDate2 AS TIMESTAMP)) AS Year,
    EXTRACT(MONTH FROM CAST(RecordDate2 AS TIMESTAMP)) AS Month,
    EXTRACT(DAY FROM CAST(RecordDate2 AS TIMESTAMP)) AS Day, 
    Date_Format(RecordDate2, 'HH:mm:ss') AS Watch_Time
FROM bright_tv.default.bright_tv_viewership;

SELECT 
    
    -- 2. Cleaned, standardized Timestamp column
    CAST(RecordDate2 AS TIMESTAMP) AS cleaned_timestamp,
    
    -- 3. Extract the Full Month Name (e.g., "January")
    date_format(CAST(RecordDate2 AS TIMESTAMP), 'MMMM') AS month_name,
     
    -- 5. Extract the Year (e.g., 2016)
    year(CAST(RecordDate2 AS TIMESTAMP)) AS extracted_year,
    
    -- 6. Extract the Day of the Month (e.g., 4)
    day(CAST(RecordDate2 AS TIMESTAMP)) AS extracted_day,
    
    -- 7. Extract the Time Only (e.g., "18:09:00")
    date_format(CAST(RecordDate2 AS TIMESTAMP), 'HH:mm:ss') AS extracted_time
FROM bright_tv.default.bright_tv_viewership;

----------------VIEWING TIME BUCKET--------------------
SELECT COUNT(Channel_Watched) AS Total_Viewing_Per_Day,
    CASE
        WHEN Hour_Of_Day BETWEEN 0 AND 5 THEN 'Late Night'
        WHEN Hour_Of_Day BETWEEN 6 AND 11 THEN 'Morning'
        WHEN Hour_Of_Day BETWEEN 12 AND 17 THEN 'Afternoon'
        WHEN Hour_Of_Day BETWEEN 18 AND 23 THEN 'Prime Time'
    END AS Time_Of_Day
FROM bright_tv.default.bright_tv_final_table
GROUP BY
    CASE
        WHEN Hour_Of_Day BETWEEN 0 AND 5 THEN 'Late Night'
        WHEN Hour_Of_Day BETWEEN 6 AND 11 THEN 'Morning'
        WHEN Hour_Of_Day BETWEEN 12 AND 17 THEN 'Afternoon'
        WHEN Hour_Of_Day BETWEEN 18 AND 23 THEN 'Prime Time'
    END
ORDER BY Time_Of_Day DESC;
----------------DURATION------------------
SELECT `Duration 2`,
FROM bright_tv.default.bright_tv_viewership;

-- Formatted duration timestamp
SELECT `Duration 2`,
    date_format(CAST(`Duration 2` AS TIMESTAMP), 'HH:mm:ss') AS timestamp_formatted_duration
FROM bright_tv.default.bright_tv_viewership;

------------------DURATION-------------
SELECT
 COALESCE(UserID0, userid4) AS UserID,
    Channel2 AS Channel_Watched,
    `Duration 2` AS Duration,
CASE
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) < 5 THEN '(0-5) Low'
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) BETWEEN 5 AND 15 THEN '5-15 Moderate'
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) BETWEEN 15 AND 30 THEN '15-30 High'
    ELSE '30+ min Very High'
    END AS Duration_Bucket
FROM bright_tv.default.bright_tv_viewership;
--------------DATA CLEANING--------------

SELECT 
    COUNT(*) AS Count,
    COUNT(DISTINCT Channel2) AS Distinct_Number_Of_Channels,
     CASE 
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet'
        ELSE Channel2 
    END AS Channel_Name
FROM bright_tv.default.bright_tv_viewership
GROUP BY Channel_Name;

--Replacing mispelled channel names with correct ones
 SELECT DISTINCT
      CASE 
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet' ELSE Channel2 END Channel_Name
FROM bright_tv.default.bright_tv_viewership;       

--The below SYNTAX returns corrected list of channel names and grouped them into DISTINCT CHANNELS to those 
--that were already correct
SELECT
    CASE
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet' ELSE Channel2 
        END Channel_Name,

    COUNT(*) AS Total_Viewership

FROM bright_tv.default.bright_tv_viewership

GROUP BY
    CASE
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet' ELSE Channel2
    END
ORDER BY Total_Viewership DESC;

 SELECT DISTINCT
      CASE 
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet' ELSE Channel2 END Channel_Name
FROM bright_tv.default.bright_tv_viewership;

WITH base AS (

SELECT COALESCE(UserID0,userid4) AS UserID
FROM bright_tv.default.bright_tv_viewership

),
processing AS (

SELECT 
    COALESCE(UserID0,userid4) AS UserID,
    DATE_FORMAT(RecordDate2, 'yyyyMM') AS Month_Id,
    DAY(RecordDate2) AS Viewing_Day,
    DATE_FORMAT(RecordDate2, 'EEEE') AS Day_Of_Week,

    CASE 
        WHEN Day_Of_Week IN ('Saturday', 'Sunday') THEN 'Weekend'
        ELSE 'Weekday'
        END AS Day_Type,

    MONTHNAME(RecordDate2) AS Watch_Month, 
  
    TO_DATE(RecordDate2) AS Watch_date,
   
CASE 
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet' ELSE Channel2 
        END Channel_Watched,

        Date_Format(RecordDate2, 'HH:mm:ss') AS Watch_Time,
        HOUR(RecordDate2) AS Hour_of_Day,
        Date_Format(`Duration 2`, 'HH:mm:ss') AS Duration

FROM bright_tv.default.bright_tv_viewership
)
SELECT*
FROM processing;


WITH user_profiles AS (
SELECT 

    TRIM(CAST(UserID AS STRING)) AS UserID,

    CASE
        WHEN TRIM(Province) = '' THEN 'Uncategorised'
        WHEN Province ILIKE 'None' THEN 'Uncategorised'
        WHEN Province ILIKE 'Uncategorised' THEN 'Uncategorised'
        ELSE Province
    END AS Region,


    Age,


    CASE
        WHEN Age BETWEEN 0 AND 2 THEN 'Infants'
        WHEN Age BETWEEN 3 AND 12 THEN 'Kids'
        WHEN Age BETWEEN 13 AND 19 THEN 'Teens'
        WHEN Age BETWEEN 20 AND 35 THEN 'Young Adults'
        WHEN Age BETWEEN 36 AND 50 THEN 'Adults'
        WHEN Age BETWEEN 51 AND 64 THEN 'Mature Adults'
        WHEN Age >= 65 THEN 'Seniors'
        ELSE 'Unknown'
    END AS Age_Bucket,


    CASE
        WHEN TRIM(Race) = '' THEN 'Unknown'
        WHEN Race = 'black' THEN 'Black'
        WHEN Race = 'white' THEN 'White'
        WHEN Race = 'indian_asian' THEN 'Indian_Asian'
        WHEN Race = 'coloured' THEN 'Coloured'
        WHEN Race ILIKE 'None' THEN 'Unknown'
        WHEN Race ILIKE 'Unknown' THEN 'Unknown'
        When Race ILIKE 'Other' THEN 'Unknown'
        ELSE Race
    END AS Race,


    CASE
        WHEN TRIM(Gender) = '' THEN 'Unknown'
        WHEN Gender = 'female' THEN 'Female'
        WHEN Gender = 'male' THEN 'Male'
        WHEN Gender ILIKE 'None' THEN 'Unknown'
        WHEN Gender ILIKE 'Unknown' THEN 'Unknown'
        ELSE Gender
    END AS Gender,

    CASE
         WHEN Email IS NOT NULL OR Email = '' OR Email NOT IN ('None') THEN 1 
         ELSE 0
        END AS email_flag,
       
    CASE
         WHEN `Social Media Handle` IS NOT NULL OR `Social Media Handle` = '' OR `Social Media Handle` NOT IN ('None') THEN 1 
         ELSE 0
        END AS SocialMediaHamdle_flag
FROM bright_tv.default.bright_tv_user_profiles
),
viewership AS (
    SELECT 

    TRIM(CAST(COALESCE(UserID0,userid4) AS STRING)) AS UserID,
    DATE_FORMAT(RecordDate2,'yyyyMM') AS Month_ID,
    DAY(RecordDate2) AS Day_Of_Month,
    DATE_FORMAT(RecordDate2,'EEEE') AS Day_Of_Week,

    CASE

        WHEN DATE_FORMAT(RecordDate2,'EEEE') IN ('Saturday','Sunday') THEN 'Weekend'
        ELSE 'Weekday'
    END AS Day_Type,
    MONTHNAME(RecordDate2) AS Watch_Month,
    TO_DATE(RecordDate2) AS Watch_Date,

    CASE
        WHEN Channel2 = 'Supersport Live Events' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'Live on SuperSport' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'DStv Events 1' THEN 'SuperSport Live Events'
        WHEN Channel2 = 'SawSee' THEN 'Sawsee'
        WHEN Channel2 = 'Break in transmission' THEN 'Break In Transmission'
        WHEN Channel2 = 'kykNET' THEN 'KyKnet'
        ELSE Channel2
    END AS Channel_Watched,
    DATE_FORMAT(RecordDate2,'HH:mm:ss') AS Watch_Time,
    HOUR(RecordDate2) AS Hour_Of_Day,
    DATE_FORMAT(`Duration 2`,'HH:mm:ss') AS Duration,

CASE
        WHEN Hour_Of_Day BETWEEN 0 AND 5 THEN 'Late Night'
        WHEN Hour_Of_Day BETWEEN 6 AND 11 THEN 'Morning'
        WHEN Hour_Of_Day BETWEEN 12 AND 17 THEN 'Afternoon'
        WHEN Hour_Of_Day BETWEEN 18 AND 23 THEN 'Prime Time'
    END AS Time_Of_Day,

CASE
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) < 5 THEN '(0-5) Low'
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) BETWEEN 5 AND 15 THEN '5-15 Moderate'
    WHEN (HOUR(Duration)*60 + MINUTE(Duration)) BETWEEN 15 AND 30 THEN '15-30 High'
    ELSE '30+ min Very High'
    END AS Duration_Bucket

FROM bright_tv.default.bright_tv_viewership
)
SELECT
    v.*,
    u.Region,
    u.Age,
    u.Age_Bucket,
    u.Race,
    u.Gender
FROM viewership AS v
LEFT JOIN user_profiles AS u
ON v.UserID = u.UserID;


```
