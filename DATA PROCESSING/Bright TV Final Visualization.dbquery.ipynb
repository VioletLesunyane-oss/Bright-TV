```sql
-----------------------TABLE OVERVIEW--------------------
SELECT *
FROM bright_tv.default.bright_tv_viewership_data_analysis;
---------------------------------------------------------
          --KEY PRFORMANCE INDICATORS (KPIs)--
---------------------------------------------------------
--Total Number of Subscribers
SELECT 
    COUNT(*) AS Number_Of_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--Returned 10000 Subscribers

--Total Number of Active Subscribers
SELECT
    COUNT(DISTINCT UserID) AS Active_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--Returned 4386 Active Subscribers

--OVERALL TABLE OF ALL SUBS AND ACTIVE SUBS
SELECT
    COUNT(UserID) AS Total_Users,
    COUNT(DISTINCT CASE 
    WHEN Channel_Watched IS NOT NULL THEN UserID 
END) AS Active_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis;

--Activation Rate
SELECT
ROUND(
    COUNT(DISTINCT CASE 
        WHEN Channel_Watched IS NOT NULL THEN UserID 
    END) * 100.0
    /
    COUNT(UserID),
2
) AS Activation_Rate_Percentage
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--43.86% of BrightTV subscribers actively consumed content during the reporting period.
--Meaning 56.14% of our Subscribers remain inactive

--NUMBER OF CHANNELS
SELECT Channel_Watched,
    COUNT(DISTINCT Channel_Watched) AS Number_Of_Channels
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched;
--Returns 17 Channels

--Tota Viewing Sessions
SELECT
ROUND(
SUM(HOUR(Duration)*60 + MINUTE(Duration) + SECOND(Duration)/60),
2
) AS Total_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis;

--Average Watch Minutes
SELECT
    ROUND(
    AVG(HOUR(Duration)*60 + MINUTE(Duration) + SECOND(Duration)/60),
    2) AS Average_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--On average, every viewing session lasts approximately 9.14 minutes.

----------------------------------------------------------------------------
                              --VISUALS--
----------------------------------------------------------------------------

----------------------------------------------------------------------------
                      --HOW IS THE BUSINESS DOING?--
----------------------------------------------------------------------------

--VIEWERSHIP BEHAVIOUR OVERTIME
SELECT
    Watch_Month,
    Gender,
    COUNT(DISTINCT UserID) AS Active_Viewers,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Watch_Month,
    Gender
ORDER BY
    Watch_Month,
    Total_Views DESC;

-------------------------------------------------------------------------------------
                  --WHICH CONTENT ARE WE SELLING TO OUR CLIENTS?--
-------------------------------------------------------------------------------------
--VIEWERSHIP BY CHANNEL
SELECT
    Channel_Watched,
    COUNT(*) AS Views
FROM bright_tv.default.bright_tv_viewership_data_analysis 
GROUP BY Channel_Watched
ORDER BY Views DESC;

--TOP 3 MOST WATCHED CHANNELS
SELECT
    Channel_Watched,
    COUNT(*) AS Views
FROM bright_tv.default.bright_tv_viewership_data_analysis 
GROUP BY Channel_Watched
ORDER BY Views DESC
LIMIT 5;

--TOP 3 LEAST WATCHED CHANNELS
SELECT
    Channel_Watched,
    COUNT(DISTINCT UserID) AS Least_Watched
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Channel_Watched
ORDER BY
    Least_watched ASC
LIMIT 3;


-----------------------------------------------------------------------------
                    --WHO ARE WE SELLING OUR PRODUCTS TO?--
-----------------------------------------------------------------------------
--VIEWERSHIP BY AGE_BUCKET
--Viewership by Age_Bucket (All Subscribers)
SELECT
    Age_Bucket,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Age_Bucket
ORDER BY
    Total_Views DESC;
--Viewership by Age_Bucket (Active Subs)   
SELECT
    Age_Bucket,
    COUNT(DISTINCT UserID) AS Total_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Age_Bucket
ORDER BY Total_Viewers DESC;
--NOTE: Young Adults consume our content the most
--TOP 3 Viewers by Age_Bucket
SELECT
    Age_Bucket,
    COUNT(DISTINCT UserID) AS Total_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Age_Bucket
ORDER BY Total_Viewers DESC
LIMIT 3;

-------------------------------------------------------------------
                --WHICH RACE DO WE CATER TO?--
-------------------------------------------------------------------
--VIEWERSHIP BY RACE
SELECT
    Race,
    COUNT(DISTINCT UserID) AS Total_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Race
ORDER BY Total_Viewers DESC;

--------------------------------------------------------------------
              --WHERE ARE OUR VIEWERS LOCATED?--
--------------------------------------------------------------------
--VIEWERSHIP BY PROVINCE
--Total Number of Subscribers
SELECT
    Region,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Total_Views DESC;

--Total Number of Active Subscribers by Province
SELECT
    Region,
    COUNT(DISTINCT UserID) AS Total_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Total_Viewers DESC;

-------------------------------------------------------------------------
                --WHICH GENDER DO WE MOST APPEAL TO?
-------------------------------------------------------------------------
--VIEWERSHIP DISTRIBUTION BY GENDER
SELECT
    Gender,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Gender
ORDER BY Total_Views DESC;
--Males are our lagest audience
--Viewership Distribution TABLE
SELECT
    Gender,
    COUNT(*) AS Total_Views,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS Percentage_Of_Viewership
FROM bright_tv.default.bright_tv_viewership_data_analysis
WHERE Gender IN ('Male', 'Female')
GROUP BY
    Gender
ORDER BY
    Total_Views DESC;

-----------------------------------------------------------------
        --WHEN IS OUR CONTENT MOST WATCHED?--
----------------------------------------------------------------
--VIEWERSHIP BY TIME_OF_DAY

SELECT COUNT(Channel_Watched) AS Total_Viewing_Per_Day,
    CASE
        WHEN Hour_Of_Day BETWEEN 0 AND 5 THEN 'Late Night'
        WHEN Hour_Of_Day BETWEEN 6 AND 11 THEN 'Morning'
        WHEN Hour_Of_Day BETWEEN 12 AND 17 THEN 'Afternoon'
        WHEN Hour_Of_Day BETWEEN 18 AND 23 THEN 'Prime Time'
    END AS Time_Of_Day
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    CASE
        WHEN Hour_Of_Day BETWEEN 0 AND 5 THEN 'Late Night'
        WHEN Hour_Of_Day BETWEEN 6 AND 11 THEN 'Morning'
        WHEN Hour_Of_Day BETWEEN 12 AND 17 THEN 'Afternoon'
        WHEN Hour_Of_Day BETWEEN 18 AND 23 THEN 'Prime Time'
    END
ORDER BY Time_Of_Day DESC;

--VIEWERSHIP BY DAY & TIME
SELECT
    Time_Of_Day,
    CASE
        WHEN Time_Of_Day = 'Late Night' THEN '00:00 - 05:59'
        WHEN Time_Of_Day = 'Morning' THEN '06:00 - 11:59'
        WHEN Time_Of_Day = 'Afternoon' THEN '12:00 - 17:59'
        WHEN Time_Of_Day = 'Prime Time' THEN '18:00 - 23:59'
    END AS Time_Range,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Time_Of_Day
ORDER BY
    Total_Views DESC;

SELECT
    Time_Of_Day,
    CASE
        WHEN Time_Of_Day = 'Late Night' THEN '00:00 - 05:59'
        WHEN Time_Of_Day = 'Morning' THEN '06:00 - 11:59'
        WHEN Time_Of_Day = 'Afternoon' THEN '12:00 - 17:59'
        WHEN Time_Of_Day = 'Prime Time' THEN '18:00 - 23:59'
    END AS Time_Range,
    COUNT(*) AS Total_Views,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS Percentage_Of_Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Time_Of_Day
ORDER BY Total_Views DESC;
--------------------------------------------------------------------------
     --ON WHICH DAY OF THE WEEK DO WE GET MOST AND LEAST VIEWS?--
--------------------------------------------------------------------------
--HIGHEST AND LOWEST VIEWERSHIP BY DAY_OF_WEEK
SELECT 
    Day_Of_Week,
    COUNT(DISTINCT UserID) AS Highest_Viewership
FROM bright_tv.default.bright_tv_final_table
GROUP BY Day_Of_Week
ORDER BY Day_Of_Week DESC;

-------------------------------------------------------------------------
                --COMPARING WEEKDAYS WITH WEEKENDS--
-------------------------------------------------------------------------
SELECT
    Day_Type,
    COUNT(*) AS Views
FROM bright_tv.default.bright_tv_final_table
GROUP BY Day_Type
ORDER BY Views DESC;
--Viewers watch our content mostly on weekdays

--------------------------------------------------------------------------
                         --DATA FILTERING--
--------------------------------------------------------------------------
--VIWERSHIP BY AGE_BUCKET FILTERED BY GENDER
SELECT
    Age_Bucket,
    Gender,
    COUNT(DISTINCT UserID) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Age_Bucket,
   Gender
ORDER BY
    Age_Bucket,
    Total_Views DESC;
--DAY_TYPE VIEWESHIP FILTERED BY GENDER
SELECT
    Day_Type,
    Gender,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Day_Type,
   Gender
ORDER BY
    Day_Type,
    Total_Views DESC;

--VIEWESHIP BY AGE_BUCKET FILTERED BY DAY_TYPE

SELECT
    Race,
    Day_Type,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Race,
   Day_Type
ORDER BY
    Race,
    Total_Views DESC;

--VIEWESHIP BY AGE_BUCKET FILTERED BY DAY_TYPE
SELECT
    Age_Bucket,
    Day_Type,
    COUNT(*) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Age_Bucket,
   Day_Type
ORDER BY
    Age_Bucket,
    Total_Views DESC;

--VIWERSHIP BY CHANNEL_WATCHED FILTERED BY GENDER

SELECT
    Channel_Watched,
    Gender,
    COUNT(DISTINCT UserID) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Channel_Watched,
   Gender
ORDER BY
    Channel_Watched,
    Total_Views DESC;
--VIEWERSHIP BY PROVINCE FILTERED BY RACE
SELECT
    Region,
    Race,
    COUNT(DISTINCT UserID) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Region,
   Race
ORDER BY
    Region,
    Total_Views DESC;

--VIEWERSHIP BY PROVINCE FILTERED BY GENDER
SELECT
    Region,
    Gender,
    COUNT(DISTINCT UserID) AS Total_Views
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
   Region,
   Gender
ORDER BY
    Region,
    Total_Views DESC;

-----------------------AVERAGE VIEWS PER SUBSCRIBER-------------
--MONTHLY VIEWER ENGAGEMENT
SELECT
    Watch_Month,
    COUNT(DISTINCT UserID) AS Active_Viewers,
    COUNT(*) AS Total_Views,
    ROUND(COUNT(*) / COUNT(DISTINCT UserID), 2) AS Avg_Views_Per_Subscriber
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Watch_Month
ORDER BY Watch_Month;
--NOTE:We only have 3 months worth of data
--VIEWER ENGAGEMENT
SELECT
ROUND(
    COUNT(*)
    / COUNT(DISTINCT UserID),2) AS Avg_Views_Per_Subscriber
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--Meaning on average a subcriber has watched BrightTV content atleast twice

---------------------------------------------------------------------------
            --EXECUTIVE VIEWERSHIP DASHBOARD AND REPORTING--
---------------------------------------------------------------------------
              --KEY PRFORMANCE INDICATORS (KPIs)--
---------------------------------------------------------------------------
--TOTAL NUMBER OF SUBSCRIBERS
SELECT 
    COUNT(*) AS Number_Of_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--Returned 10000 Subscribers

--TOTAL NUMBER OF ACTIVE SUBSCRIBERS
SELECT
    COUNT(DISTINCT UserID) AS Active_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--Returned 4386 Active Subscribers

--OVERALL TABLE OF ALL SUBS AND ACTIVE SUBS
SELECT
    COUNT(UserID) AS Total_Users,
    COUNT(DISTINCT CASE 
    WHEN Channel_Watched IS NOT NULL THEN UserID 
END) AS Active_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis;

--ACTIVATION RATE
SELECT
ROUND(
COUNT(DISTINCT UserID)*100.0/10000,
2
) AS Activation_Rate
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--43.86% of BrightTV subscribers actively consumed content during the reporting period.
--Meaning 56.14% of our Subscribers remain inactive

--NUMBER OF CHANNELS
SELECT Channel_Watched,
    COUNT(DISTINCT Channel_Watched) AS Number_Of_Channels
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched;
--Returns 17 Channels

--TOTAL VIEWING SESSION IN MINUTES
SELECT
ROUND(
SUM(HOUR(Duration)*60 + MINUTE(Duration) + SECOND(Duration)/60),
2
) AS Total_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis;

--AVERAGE VIEWING MINUTES
SELECT
    ROUND(
    AVG(HOUR(Duration)*60 + MINUTE(Duration) + SECOND(Duration)/60),
    2) AS Average_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis;
--On average, every viewing session lasts approximately 9.14 minutes.

----------------------------------------------------------------------------
                            --VIEWER DEMOGRAPHICS--
----------------------------------------------------------------------------
--GENDER DISTRIBUTION
SELECT
    Gender,
    COUNT(DISTINCT UserID) AS Active_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Gender
ORDER BY Active_Subscribers DESC;

--AGE GROUP
SELECT
    Age_Bucket,
    COUNT(DISTINCT UserID) AS Active_Subscribers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Age_Bucket
ORDER BY Active_Subscribers DESC;

--RACE
SELECT
    Race,
    COUNT(DISTINCT UserID) AS Active_Subs_By_Race
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Race
ORDER BY Active_Subs_By_Race DESC;

----------------------------------------------------------------------------
                        --GEOGRAPHIC ANALYSIS--
----------------------------------------------------------------------------
--PROVINCE DISTRIBUTION
--Active Subscribers by Province
SELECT
    Region,
    COUNT(DISTINCT UserID) AS Active_Subs_By_Region
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Active_Subs_By_Region DESC;

--Watch Minutes by Province
SELECT
    Region,
    ROUND(
    SUM(HOUR(Duration)*60 + MINUTE(Duration)+SECOND(Duration)/60),2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Watch_Minutes DESC;

--Average Watch Minutes by Province

SELECT
    Region,
    ROUND(
    AVG(HOUR(Duration)*60 + MINUTE(Duration)+SECOND(Duration)/60),2) AS Avg_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Avg_Watch_Minutes DESC;

---------------------------------------------------------------------------
                           --VIEWER BEHAVIOUR--
---------------------------------------------------------------------------
--VIEWING BY HOUR
SELECT
    Hour_Of_Day,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Hour_Of_Day
ORDER BY Hour_Of_Day;

--VIEWING BY DAY OF WEEK
SELECT
    Day_Of_Week,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Day_Of_Week;

--VIEWING BY DAY_OF_WEEK AND HOUR_OF_DAY (Heat Map)
SELECT
    Day_Of_Week,
    Hour_Of_Day,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Day_Of_Week,
    Hour_Of_Day;
--VIEWING BY TIME OF DAY
SELECT
Time_Of_Day,
COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Time_Of_Day
ORDER BY Viewing_Sessions DESC;

--MONTHLY VIEWING TREND
SELECT
Watch_Month,
COUNT(*) AS Viewing_Sessions,
COUNT(DISTINCT UserID) AS Active_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Watch_Month
ORDER BY Watch_Month;

--DURATION DISTRIBUTION
SELECT
    CASE
        WHEN (HOUR(Duration)*60+MINUTE(Duration)) <5 THEN '0-5 Minutes'
        WHEN (HOUR(Duration)*60+MINUTE(Duration)) BETWEEN 5 AND 10 THEN '5-10 Minutes'
        WHEN (HOUR(Duration)*60+MINUTE(Duration)) BETWEEN 11 AND 20 THEN '11-20 Minutes'
        ELSE '20+ Minutes'
        END AS Duration_Bucket,
        COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Duration_Bucket
ORDER BY Viewing_Sessions DESC;

----------------------------------------------------------------------------
                           --CHANNEL PERFORMANCE--
----------------------------------------------------------------------------
--TOP 10 CHANNELS
SELECT
Channel_Watched,
COUNT(DISTINCT UserID) AS Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
ORDER BY Viewers DESC
LIMIT 10;

--TOP 5 LEAST WATCHED CHANNELS
SELECT
    Channel_Watched,
    COUNT(DISTINCT UserID) AS Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
ORDER BY Viewers ASC
LIMIT 5;

--WATCH MINUTES BY CHANNEL
SELECT
    Channel_Watched,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
ORDER BY Watch_Minutes DESC;

--AVERAGE WATCH MINUTES BY CHANNEL

SELECT
    Channel_Watched,
    ROUND(
    AVG(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2) AS Avg_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
ORDER BY Avg_Watch_Minutes DESC;

---------------------------------------------------------------------------
                        --CONSUMPTION DRIVERS--
--------------------------------------------------------------------------
--AGE VS WATCH MINUTES
SELECT
    Age_Bucket,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Age_Bucket
ORDER BY Watch_Minutes DESC;

--GENDER VS VIEWING MINUTES
SELECT
    Gender,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Gender
ORDER BY Watch_Minutes DESC;

--CHANNEL BY TIME_OF_DAY
SELECT
    Time_Of_Day,
    Channel_Watched,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Time_Of_Day,
        Channel_Watched;

--PROVINCE VS VIEWING MINUTES
SELECT
    Region,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2
    ) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region
ORDER BY Watch_Minutes DESC;

--PROVINCE BY AGE_BUCKET
SELECT
    Region,
    Age_Bucket,
    COUNT(DISTINCT UserID) AS Viewers,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60
    ),2
    ) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Region,
         Age_Bucket;
--TIME BUCKET VS VIEWING MINUTES
SELECT
    Time_Of_Day,
    ROUND(
    SUM(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Time_Of_Day
ORDER BY Watch_Minutes DESC;

---------------------------------------------------------------------------
                      --GROWTH OPPORTUNITIES--
---------------------------------------------------------------------------
--LOWEST PERFORMING CHANNELS
SELECT
    Channel_Watched,
    COUNT(DISTINCT UserID) AS Active_Viewers
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
ORDER BY Active_Viewers ASC;

--LOWEST VIEWING DAYS
SELECT
    Day_Of_Week,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Day_Of_Week
ORDER BY Viewing_Sessions ASC;

--LOWEST VIEWING TIME BUCKET
SELECT
    Time_Of_Day,
    COUNT(*) AS Viewing_Sessions
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Time_Of_Day
ORDER BY Viewing_Sessions ASC;

--CHANNEL ENGAGEMENT SCORE
WITH Channel_Metrics AS (
SELECT
    Channel_Watched,
    COUNT(DISTINCT UserID) AS Unique_Viewers,

    ROUND(
        SUM(
            HOUR(Duration) * 60 +
            MINUTE(Duration) +
            SECOND(Duration) / 60
        ),2
    ) AS Total_Watch_Minutes,

    ROUND(
        AVG(
            HOUR(Duration) * 60 +
            MINUTE(Duration) +
            SECOND(Duration) / 60
        ),2
    ) AS Avg_Watch_Minutes

FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
)
SELECT
    Channel_Watched,
    Unique_Viewers,
    Total_Watch_Minutes,
    Avg_Watch_Minutes,

    ROUND(
        (Unique_Viewers * 0.40) +
        (Total_Watch_Minutes * 0.40) +
        (Avg_Watch_Minutes * 0.20),
    2) AS Engagement_Score
FROM Channel_Metrics
ORDER BY Engagement_Score DESC;

--PROVINCE x AGE BUCKET MATRIX
SELECT
    Region,
    Age_Bucket,
    COUNT(DISTINCT UserID) AS Active_Viewers,
    ROUND(
        SUM(
            HOUR(Duration)*60 +
            MINUTE(Duration) +
            SECOND(Duration)/60
        ),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Region,
    Age_Bucket
ORDER BY
    Region,
    Watch_Minutes DESC;

--CHANNEL x TIME OF DAY MATRIX
SELECT
    Time_Of_Day,
    Channel_Watched,
    COUNT(*) AS Viewing_Sessions,
    COUNT(DISTINCT UserID) AS Active_Viewers,
    ROUND(
        SUM(
            HOUR(Duration)*60 +
            MINUTE(Duration)+
            SECOND(Duration)/60
        ),
    2) AS Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY
    Time_Of_Day,
    Channel_Watched
ORDER BY
    Time_Of_Day,
    Watch_Minutes DESC;

WITH Channel_Performance AS (
SELECT
Channel_Watched,
COUNT(DISTINCT UserID) AS Viewers,
    ROUND(
    AVG(HOUR(Duration)*60+MINUTE(Duration)+SECOND(Duration)/60
    ),2) AS Avg_Watch_Minutes
FROM bright_tv.default.bright_tv_viewership_data_analysis
GROUP BY Channel_Watched
)
SELECT *
FROM Channel_Performance
ORDER BY
    Viewers DESC,
    Avg_Watch_Minutes DESC;
```
